```python
from transformers import pipeline
```

```python
text = """
Hey there drama. Are you into karma?
Because I am dying to believe you are next.

Oh, is your mind erasing?
Then you better chase it.
Because the thought is scaring me to death.

Can you feel it?
Because I feel it now.

You drive me crazy with that chit chat.
Do you want that?
Or do you want to take me home tonight?

All this chit chat is holding me back.
And I am breaking just to bend your light.

I thought I told you.
I really need your sugar.

A rollercoaster.
I am going supernova.
"""
pipe=pipeline("text-to-speech",model="suno/bark-small",device="cuda")
output=pipe(text)
```

```python
from IPython.display import Audio
Audio(output["audio"],rate=output["sampling_rate"])
```
